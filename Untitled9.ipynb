```python

import numpy as np
import matplotlib.pyplot as plt
import random
import math

# 1. Define the environment
START = np.array([0.0, 0.0])
GOAL = np.array([10.0, 10.0])

# Obstacles: (x, y, radius)
obstacles = [
    (4, 4, 1.5),
    (6, 7, 1.3),
    (7, 3, 1.2)
]

# PSO parameters
NUM_PARTICLES = 30
MAX_ITERATIONS = 100
NUM_WAYPOINTS = 3

W = 0.7       # Inertia weight
C1 = 1.5      # Cognitive coefficient
C2 = 1.5      # Social coefficient

DIM = NUM_WAYPOINTS * 2
LOWER_BOUND = 0
UPPER_BOUND = 10


# 2. Calculate distance between two points
def distance(p1, p2):
    return np.linalg.norm(p1 - p2)


# 3. Find minimum distance between an obstacle
#    and a line segment
def segment_distance(a, b, center):
    ab = b - a

    if np.dot(ab, ab) == 0:
        return distance(a, center)

    t = np.dot(center - a, ab) / np.dot(ab, ab)
    t = max(0, min(1, t))

    closest = a + t * ab

    return distance(closest, center)


# 4. Fitness function
def fitness(position):

    points = [START]

    # Convert particle position into waypoints
    for i in range(NUM_WAYPOINTS):
        x = position[2 * i]
        y = position[2 * i + 1]
        points.append(np.array([x, y]))

    points.append(GOAL)

    total_distance = 0
    penalty = 0

    # Calculate path length
    for i in range(len(points) - 1):

        a = points[i]
        b = points[i + 1]

        total_distance += distance(a, b)

        # Check obstacle collisions
        for ox, oy, radius in obstacles:

            center = np.array([ox, oy])

            d = segment_distance(a, b, center)

            if d < radius:
                penalty += 10000 + (radius - d) * 1000

    return total_distance + penalty


# 5. Initialize particles
particles = np.random.uniform(
    LOWER_BOUND, UPPER_BOUND,
    (NUM_PARTICLES, DIM)
)

velocities = np.random.uniform(
    -1, 1,
    (NUM_PARTICLES, DIM)
)

# Personal best
pbest = particles.copy()

pbest_fitness = np.array([
    fitness(p) for p in particles
])

# Global best
best_index = np.argmin(pbest_fitness)

gbest = pbest[best_index].copy()
gbest_fitness = pbest_fitness[best_index]

history = []


# 6. PSO main loop
for iteration in range(MAX_ITERATIONS):

    for i in range(NUM_PARTICLES):

        r1 = random.random()
        r2 = random.random()

        # Update velocity
        velocities[i] = (
            W * velocities[i]
            + C1 * r1 * (pbest[i] - particles[i])
            + C2 * r2 * (gbest - particles[i])
        )

        # Update position
        particles[i] = particles[i] + velocities[i]

        # Keep particles within the search space
        particles[i] = np.clip(
            particles[i],
            LOWER_BOUND,
            UPPER_BOUND
        )

        # Evaluate new fitness
        current_fitness = fitness(particles[i])

        # Update personal best
        if current_fitness < pbest_fitness[i]:

            pbest[i] = particles[i].copy()
            pbest_fitness[i] = current_fitness

            # Update global best
            if current_fitness < gbest_fitness:

                gbest = particles[i].copy()
                gbest_fitness = current_fitness

    history.append(gbest_fitness)


# 7. Display optimized path
path = [START]

for i in range(NUM_WAYPOINTS):
    path.append(
        np.array([gbest[2 * i], gbest[2 * i + 1]])
    )

path.append(GOAL)

print("\nROBOT PATH PLANNING USING PSO")
print("--------------------------------")

for i, point in enumerate(path):
    print(f"Point {i}: ({point[0]:.2f}, {point[1]:.2f})")

print("\nBest Fitness Value:", round(gbest_fitness, 2))

# 8. Visualize the result
plt.figure(figsize=(8, 8))

# Draw obstacles
for ox, oy, radius in obstacles:

    circle = plt.Circle(
        (ox, oy), radius,
        color='red', alpha=0.5
    )

    plt.gca().add_patch(circle)

# Extract path coordinates
x_values = [p[0] for p in path]
y_values = [p[1] for p in path]

# Draw optimized path
plt.plot(
    x_values, y_values,
    'b-o', linewidth=2,
    label='Optimized Robot Path'
)

# Mark start and goal
plt.scatter(0, 0, color='green', s=100, label='Start')
plt.scatter(10, 10, color='blue', s=100, label='Goal')

plt.xlim(0, 10)
plt.ylim(0, 10)

plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("Robot Path Planning using PSO")

plt.grid(True)
plt.legend()
plt.show()


# Plot convergence
plt.figure(figsize=(8, 4))

plt.plot(history)
plt.xlabel("Iteration")
plt.ylabel("Best Fitness")
plt.title("PSO Convergence Graph")
plt.grid(True)
plt.show()
```

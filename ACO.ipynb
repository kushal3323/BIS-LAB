```python

import math
import random

# Step 1: Define delivery locations and coordinates
cities = {
    "Warehouse": (0, 0),
    "Customer A": (2, 6),
    "Customer B": (5, 8),
    "Customer C": (8, 5),
    "Customer D": (6, 1),
    "Customer E": (1, 2)
}

city_names = list(cities.keys())
n = len(city_names)

# Step 2: Calculate distance between every pair of cities
distance = [[0] * n for _ in range(n)]

for i in range(n):
    for j in range(n):
        x1, y1 = cities[city_names[i]]
        x2, y2 = cities[city_names[j]]

        distance[i][j] = math.sqrt(
            (x2 - x1)**2 + (y2 - y1)**2
        )

# Step 3: Initialize ACO parameters
num_ants = 20
alpha = 1
beta = 5
rho = 0.5
Q = 100
iterations = 100
initial_pheromone = 1.0

random.seed(42)

# Initialize pheromone matrix
pheromone = [
    [initial_pheromone if i != j else 0
     for j in range(n)]
    for i in range(n)
]

best_route = None
best_distance = float('inf')

# Step 4: Start ACO iterations
for iteration in range(iterations):

    all_routes = []

    # Each ant constructs one route
    for ant in range(num_ants):

        # All ants start from the warehouse (index 0)
        start = 0
        route = [start]
        unvisited = set(range(1, n))

        current = start

        # Visit all remaining cities
        while unvisited:

            choices = sorted(unvisited)
            probabilities = []

            # Calculate probability weights
            for city in choices:

                pheromone_value = (
                    pheromone[current][city] ** alpha
                )

                heuristic_value = (
                    1 / distance[current][city]
                ) ** beta

                probability = (
                    pheromone_value * heuristic_value
                )

                probabilities.append(probability)

            # Normalize probabilities
            total = sum(probabilities)

            probabilities = [
                p / total for p in probabilities
            ]

            # Select next city probabilistically
            next_city = random.choices(
                choices,
                weights=probabilities,
                k=1
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)
            current = next_city

        # Return to warehouse
        route.append(start)

        # Calculate total route distance
        total_distance = 0

        for i in range(len(route) - 1):
            total_distance += distance[
                route[i]
            ][
                route[i + 1]
            ]

        all_routes.append((route, total_distance))

        # Update best solution
        if total_distance < best_distance:
            best_distance = total_distance
            best_route = route.copy()

    # Step 5: Evaporate pheromones
    for i in range(n):
        for j in range(n):
            if i != j:
                pheromone[i][j] *= (1 - rho)

    # Step 6: Deposit new pheromones
    for route, route_distance in all_routes:

        deposit = Q / route_distance

        for i in range(len(route) - 1):

            a = route[i]
            b = route[i + 1]

            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

# Step 7: Display best solution
print("ANT COLONY OPTIMIZATION - DELIVERY ROUTE")
print("----------------------------------------")

print("\nBest Delivery Route:")

best_route_names = [
    city_names[i] for i in best_route
]

print(" -> ".join(best_route_names))

print("\nMinimum Total Distance:",
      round(best_distance, 2), "units")

print("\nTotal Cities Visited:", n - 1)

print("Algorithm Iterations:", iterations)

print("Number of Ants:", num_ants)
```

```python

import math
import random

# Step 1: Take input
n = int(input("Enter number of cities: "))

cities = []

print("Enter coordinates of each city (x y):")

for i in range(n):
    x, y = map(float, input().split())
    cities.append((x, y))

# Step 2: Take ACO parameters
num_ants = int(input("Enter number of ants: "))
iterations = int(input("Enter number of iterations: "))
alpha = float(input("Enter alpha (pheromone importance): "))
beta = float(input("Enter beta (distance importance): "))
rho = float(input("Enter evaporation rate: "))
Q = float(input("Enter pheromone deposit constant Q: "))

random.seed(42)

# Step 3: Calculate distance matrix
dist = [[0] * n for _ in range(n)]

for i in range(n):
    for j in range(n):
        x1, y1 = cities[i]
        x2, y2 = cities[j]

        dist[i][j] = math.sqrt(
            (x2 - x1)**2 + (y2 - y1)**2
        )

# Step 4: Initialize pheromone matrix
pheromone = [[1.0 if i != j else 0
              for j in range(n)] for i in range(n)]

best_route = None
best_distance = float('inf')

# Step 5: Main ACO loop
for iteration in range(iterations):

    all_routes = []

    for ant in range(num_ants):

        start = 0
        route = [start]
        unvisited = set(range(1, n))
        current = start

        # Construct route
        while unvisited:

            choices = sorted(unvisited)
            weights = []

            for city in choices:

                p = pheromone[current][city] ** alpha
                h = (1 / dist[current][city]) ** beta

                weights.append(p * h)

            next_city = random.choices(
                choices, weights=weights, k=1
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)
            current = next_city

        # Return to starting city
        route.append(start)

        # Calculate route distance
        total = 0

        for i in range(len(route) - 1):
            total += dist[route[i]][route[i + 1]]

        all_routes.append((route, total))

        # Update best route
        if total < best_distance:
            best_distance = total
            best_route = route.copy()

    # Step 6: Evaporate pheromone
    for i in range(n):
        for j in range(n):
            if i != j:
                pheromone[i][j] *= (1 - rho)

    # Step 7: Deposit pheromone
    for route, total in all_routes:

        deposit = Q / total

        for i in range(len(route) - 1):
            a = route[i]
            b = route[i + 1]

            pheromone[a][b] += deposit
            pheromone[b][a] += deposit
print("\n----- ACO TSP RESULT -----")

print("Best Route:")

for i in range(len(best_route)):
    print(best_route[i] + 1, end="")

    if i < len(best_route) - 1:
        print(" -> ", end="")

print("\nMinimum Distance:", round(best_distance, 2))
print("Iterations:", iterations)
print("Ants:", num_ants)
```
